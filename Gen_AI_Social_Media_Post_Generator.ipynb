```python
import google.generativeai as genai
from google.colab import userdata

genai.configure(
    api_key=userdata.get("GEMINI_API_KEY")
)

def generate_social_post(topic, platform, tone):

    system_prompt = (
        "You are an expert social media content writer. "
        "Write platform-specific captions that match the writing style, "
        "length, hashtags, and tone of the selected social media platform."
    )

    user_prompt = f"""
    Generate a ready-to-post social media caption.

    Topic: {topic}

    Platform: {platform}

    Tone: {tone}

    Requirements:

    If platform is LinkedIn:
    - Around 120 words
    - Professional style
    - 3-5 hashtags
    - End with a question

    If platform is Twitter/X:
    - Maximum 280 characters
    - Short and engaging
    - 2-3 hashtags

    If platform is Instagram:
    - Friendly and engaging
    - Around 80-120 words
    - Use emojis
    - 5-8 hashtags

    If platform is Facebook:
    - Conversational
    - Around 100 words
    - 2-4 hashtags

    Return only the caption.
    """

    model = genai.GenerativeModel(
        model_name="gemini-3.5-flash-lite",
        system_instruction=system_prompt
    )

    response = model.generate_content(
        user_prompt,
        generation_config={
            "temperature":1.0,
            "max_output_tokens":300,
        }
    )

    return response.text
```

```python
topic = input("Enter the topic: ")
platform = input("Enter platform (LinkedIn/Twitter/X/Instagram/Facebook): ")
tone = input("Enter tone: ")

post = generate_social_post(topic, platform, tone)

print("\nGenerated Social Media Post:\n")
print(post)
```

```python
filename = platform.replace("/", "_") + "_" + topic.replace(" ", "_") + ".txt"

file = open(filename, "w")

file.write("Topic: " + topic + "\n")
file.write("Platform: " + platform + "\n")
file.write("Tone: " + tone + "\n\n")
file.write(post)

file.close()

print("File saved successfully.")
```
